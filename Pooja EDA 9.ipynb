```python
df.isnull().sum()

df = df.dropna()

df
```

```python
print("Zero Bedrooms:", (df["bedrooms"] == 0).sum())
print("Zero Bathrooms:", (df["bathrooms"] == 0).sum())
```

```python
df["bedrooms"] = df["bedrooms"].replace(
    0, df["bedrooms"].replace(0, pd.NA).median()
)

df["bathrooms"] = df["bathrooms"].replace(
    0, df["bathrooms"].replace(0, pd.NA).median()
)

df[["bedrooms", "bathrooms"]]
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    format="%Y%m%dT%H%M%S"
)

df["date"]
```

```python
df["sale_year"] = df["date"].dt.year

df[["date", "sale_year"]]
```

```python
df["property_age"] = df["sale_year"] - df["yr_built"]

df[["yr_built", "sale_year", "property_age"]]
```

```python
df["is_renovated"] = df["yr_renovated"].apply(
    lambda x: 1 if x > 0 else 0
)

df[["yr_renovated", "is_renovated"]]
```

```python
df["price_per_sqft_living"] = (
    df["price"] / df["sqft_living"]
)

df[["price", "sqft_living", "price_per_sqft_living"]]
```

```python
df["price_per_sqft_lot"] = (
    df["price"] / df["sqft_lot"]
)

df[["price", "sqft_lot", "price_per_sqft_lot"]]
```
